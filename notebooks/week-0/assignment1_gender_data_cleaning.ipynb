# **Cleaning Gender using Mapping to integers**

Genders can be mapped unto various integers. This has the advatage of being easier and faster to do opertations on compared to objects.

Here we will map male as 1, female as 0 and other as 2.
Those are the only categories presEnt in the data currently.

```python
gender_map = {
    '0': 0,
    'Male': 1,
    'Female': 0,
    'FEMALE': 0,
    '1': 1,
    'MALE': 1
}

df_clean =  df_raw.copy()
df_clean['Gender'] = df_raw['Gender'].map(gender_map)
df_clean = df_clean.fillna(2) #fills all categroies that dont match the male/female formats with 2.

df_clean.head(10)
```

```python
print(f"Number of Null Values Left: {df_clean['Gender'].isnull().sum()}")
#all existing values in the dataset have been mapped thus leaving no null values
```

# **Cleaning Gender using Dummy Variables**

The previous data cleaning implementation had the problem that it required us to map genders unto arbitray integers. This makes debugging more difficult and makes the mapping inherently subjective.

In order to avoid this, Dummy Variables can be used instead in order to encode all existing categories within the gender column into unique booleans for each class.

```python
#we first need to map the values in order to standardize the format the genders comein.
gender_encoder_map = {
    '0': 'Female',
    'Male': 'Male',
    'Female': 'Female',
    'FEMALE': 'Female',
    '1': 'Male',
    'MALE': 'Male'
}

df_encoded =  df_raw.copy()
df_encoded['Gender'] = df_raw['Gender'].map(gender_encoder_map)
df_encoded = df_encoded.fillna('Other')

df_encoded.head(10)
```

```python
encoded_unique_genders = list(df_encoded.Gender.unique())
print(f"Encoded Unique genders: {encoded_unique_genders}")
```

```python
df_clean_encoded = pd.get_dummies(df_encoded, columns=['Gender']) #creates column for each category. however, they come as booleans.
df_clean_encoded.head(10)
```

```python
df_clean_encoded['Gender_Female'] = df_clean_encoded['Gender_Female'].astype(int) #casts the boolean to integer seeing as that was the data type used in tutorial
df_clean_encoded['Gender_Male'] = df_clean_encoded['Gender_Male'].astype(int) #casts the boolean to integer seeing as that was the data type used in tutorial
#df_clean_encoded['Gender_Other'] = df_clean_encoded['Gender_Other'].astype(int) #casts the boolean to integer seeing as that was the data type used in tutorial

df_clean_encoded.head(10)
#data has been successfully mapped
```
